```python
import pyodbc

conn = pyodbc.connect(                         
    "driver={ODBC Driver 17 for SQL Server};"
    "server=localhost\\SQLEXPRESS;"
    "database=ETL_Sales;"
    "trusted_connection=yes;" 
)

cursor = conn.cursor()
```

```python
import pandas as pd
df = pd.read_csv('sales.csv')
df
```

```python
cursor.execute( """
IF OBJECT_ID('Sales', 'U') IS NOT NULL DROP TABLE Sales;
CREATE TABLE Sales (
    order_id INT,
    customer VARCHAR(50),
    product VARCHAR(50),
    quantity INT,
    unit_price FLOAT,
    total_amount FLOAT 
)
""")
```

```python
sales_list = df.values.tolist()
sales_list
```

```python
cursor.executemany(
    "INSERT INTO Sales VALUES (?,?,?,?,?,?)",
    sales_list
)

conn.commit()
```

```python
df = pd.read_sql_query('SELECT * FROM Sales', conn)

df
```

```python
cursor.close()
conn.close()
```
